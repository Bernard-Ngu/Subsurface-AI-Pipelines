```python
def plot_smlp(df, well_name):
    # Ensure data is sorted by DEPTH (Stratigraphic order)
    # Using 'DEPT' or 'DEPTH' depending on your LAS file
    df_sorted = df.sort_values(by='DEPTH') 
    
    # Handle missing values in Porosity/Permeability
    df_clean = df_sorted.dropna(subset=['CORE-POROSITY', 'C0RE-PERM']) # Replace with your curve names
    
    # Calculate Flow Capacity (Kh) and Storage Capacity (Ph)
    # Assuming constant sampling rate, h is constant, so we just sum Perm and Poro
    kh = df_clean['C0RE-PERM'].values
    ph = df_clean['CORE-POROSITY'].values
    
    # Calculate Cumulative Sums
    cum_kh = np.cumsum(kh)
    cum_ph = np.cumsum(ph)
    
    # Normalize (0 to 1)
    cum_kh_norm = cum_kh / cum_kh[-1]
    cum_ph_norm = cum_ph / cum_ph[-1]
    
    # Plot
    plt.figure(figsize=(6, 6))
    plt.plot(cum_ph_norm, cum_kh_norm, label=f'SMLP - {well_name}', linewidth=2)
    plt.plot([0, 1], [0, 1], 'k--', label='Homogeneous (Line of Equality)')
    
    plt.title(f'Stratigraphic Modified Lorenz Plot (SMLP): {well_name}')
    plt.xlabel('Normalized Cumulative Storage Capacity (Phi*h)')
    plt.ylabel('Normalized Cumulative Flow Capacity (K*h)')
    plt.legend()
    plt.grid(True)
    plt.show()

# Run for the first well in your list
plot_smlp(all_wells[0], all_wells[0]['WELL'].iloc[0])
```

```python
def plot_smlp(df, well_name):
    # Ensure data is sorted by DEPTH (Stratigraphic order)
    # Using 'DEPT' or 'DEPTH' depending on your LAS file
    df_sorted = df.sort_values(by='DEPTH') 
    
    # Handle missing values in Porosity/Permeability
    df_clean = df_sorted.dropna(subset=['PERM', 'PHIE']) # Replace with your curve names
    
    # Calculate Flow Capacity (Kh) and Storage Capacity (Ph)
    # Assuming constant sampling rate, h is constant, so we just sum Perm and Poro
    kh = df_clean['PERM'].values
    ph = df_clean['PHIE'].values     #df_clean['PHIE'].values
    
    # Calculate Cumulative Sums
    cum_kh = np.cumsum(kh)
    cum_ph = np.cumsum(ph)
    
    # Normalize (0 to 1)
    cum_kh_norm = cum_kh / cum_kh[-1]
    cum_ph_norm = cum_ph / cum_ph[-1]
    
    # Plot
    plt.figure(figsize=(6, 6))
    plt.plot(cum_ph_norm, cum_kh_norm, label=f'SMLP - {well_name}', linewidth=2)
    plt.plot([0, 1], [0, 1], 'k--', label='Homogeneous (Line of Equality)')
    
    plt.title(f'Stratigraphic Modified Lorenz Plot (SMLP): {well_name}')
    plt.xlabel('Normalized Cumulative Storage Capacity (Phi*h)', fontsize=14)
    plt.ylabel('Normalized Cumulative Flow Capacity (K*h)', fontsize=14)
    plt.legend(fontsize=14)
    plt.grid(True)
    #plt.savefig('SMLP.pdf', dpi=150)
    plt.show()

# Run for the first well in your list
plot_smlp(all_wells[0], all_wells[0]['WELL'].iloc[0])
#plt.savefig('SMLP.pdf', dpi=150)
```

```python
def calculate_vertical_variogram(df, prop_col='PERM', max_lag=50):
    """
    Calculates semivariance for vertical lags.
    max_lag: maximum number of samples to look down (e.g., 50 samples = 25ft if step is 0.5)
    """
    data = df[prop_col].dropna().values
    lags = range(1, max_lag)
    semivariance = []
    
    for h in lags:
        # Calculate difference between points separated by lag h
        diff = data[:-h] - data[h:]
        gamma = 0.5 * np.mean(diff**2)
        semivariance.append(gamma)
        
    return lags, semivariance

# Select a well and calculate
well_data = all_wells[0] # Using first well
lags, gamma = calculate_vertical_variogram(well_data, prop_col='PERM', max_lag=100)

# Convert lags to distance (assuming 0.5 ft or 0.1524 m step)
step_size = 0.1524 # Change to 0.5 if feet
lag_distance = [x * step_size for x in lags]

# Plot
plt.figure(figsize=(8, 5))
plt.plot(lag_distance, gamma, 'o-')
plt.title('Vertical Variogram (Permeability)', fontsize=16)
plt.xlabel('Vertical Lag Distance (m)', fontsize=16)
plt.ylabel('Semivariance (γ)', fontsize=16)
plt.grid(True)
#plt.savefig('vario.pdf', dpi=150)
plt.show()
```

```python
def sliding_window_cv(df, window_size=50):
    """
    window_size: Number of samples in the window (e.g., 50 samples)
    """
    # Calculate rolling Mean and Std Dev
    rolling_mean = df['PERM'].rolling(window=window_size).mean()
    rolling_std = df['PERM'].rolling(window=window_size).std()
    
    # Cv = Std / Mean
    df['Heterogeneity_Cv'] = rolling_std / rolling_mean
    return df

# Apply to one well
well_with_cv = sliding_window_cv(all_wells[0].copy(), window_size=50)

# Plotting the "Heterogeneity Log"
fig, ax = plt.subplots(nrows=1, ncols=2, figsize=(12, 10), sharey=True)



# Track 2: Permeability
ax[0].semilogx(well_with_cv['PERM'], well_with_cv['DEPTH'], color='red')
ax[0].set_xlabel('Permeability (mD)')
ax[0].grid()

# Track 3: Heterogeneity (Cv)
ax[1].plot(well_with_cv['Heterogeneity_Cv'], well_with_cv['DEPTH'], color='blue')
ax[1].set_xlabel('Heterogeneity (Cv)')
ax[1].set_title('Heterogeneity Log')
ax[1].grid()

plt.suptitle("Zonal Heterogeneity Analysis")
plt.show()
```
