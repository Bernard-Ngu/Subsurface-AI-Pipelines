```python
import matplotlib.pyplot as plt
import numpy as np

# --- 1. Map data points to FZI zones ---
phi = J_2['CORE-POROSITY']
k = J_2['C0RE-PERM']

# FZI relationship: RQI = 0.0314 * sqrt(k/phi), PHIZ = phi/(1-phi)
fzi_points = (0.0314 * np.sqrt(k / phi)) / (phi / (1 - phi))

# --- 2. Color logic as requested ---
point_colors = []
for fzi in fzi_points:
    if 1 <= fzi <= 4:
        point_colors.append('red')   # Between FZI 1 and 4
    elif 5 <= fzi <= 7:
        point_colors.append('black') # Between FZI 5 and 7
    elif fzi >= 8:
        point_colors.append('blue')  # Above FZI 8
    else:
        point_colors.append('red')  # Catch-all for points in gaps (e.g., between 4-5 or 7-8)

plt.figure(figsize=(6, 6))

# Plot scatter using the color list
plt.scatter(phi, k, color=point_colors)

# Plotting settings
plt.yscale('log')
plt.grid(visible=None, which='both', axis='both')
porosity = np.linspace(0.0001, 0.35, 50)

# Your FZI lines
plt.plot(porosity, perm1,  color='brown', label='FZI=1')
plt.plot(porosity, perm2,  color='purple', label='FZI=2')
plt.plot(porosity, perm3,  color='red', label='FZI=3')
plt.plot(porosity, perm4,  color='b', label='FZI=4')
plt.plot(porosity, perm5,  color='yellow', label='FZI=5')          
plt.plot(porosity, perm6,  color='black', label='FZI=6')
plt.plot(porosity, perm7,  color='skyblue', label='FZI=7')
plt.plot(porosity, perm8,  color='magenta', label='FZI=8')
plt.plot(porosity, perm9,  color='pink', label='FZI=9')
plt.plot(porosity, perm10, color='green', label='FZI=10')

# Formatting
plt.xlabel('Core porosity, dec.', fontdict={'fontsize':12, 'fontweight':'bold'})
plt.ylabel('Core permeability, mD', fontdict={'fontsize':12, 'fontweight':'bold'})
plt.ylim(100, 1000)
plt.xlim(0.1, .35)
plt.legend(loc='upper left', fontsize='small', ncol=2)
plt.tight_layout()
plt.savefig("p.pdf", format="pdf", bbox_inches='tight')
plt.show()
```
